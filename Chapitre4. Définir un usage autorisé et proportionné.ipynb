Pour étudier les réhospitalisations, l'équipe hésite à demander le dossier complet. Quelles colonnes lui sont réellement nécessaires ? Recherchez sur quelle base légale elle pourrait mener l'étude et quelle condition supplémentaire s'applique aux données de santé. RGPD, articles 5, 6 et 9

# Résumé du Chapitre 4 — Définir un usage autorisé et proportionné

**Contexte :** une équipe veut construire un modèle prédictif de réhospitalisations à partir de dossiers patients. Le chapitre montre comment rester légal sans céder au réflexe du « prenons toutes les données au cas où ».

## 1. La minimisation des données (Art. 5 RGPD)

- Les données doivent être **« adéquates, pertinentes et limitées à ce qui est nécessaire »** (Art. 5.1.c).
- Extraire le dossier patient complet (identité, adresse, observations libres, psychiatrie…) est **illégal et disproportionné**.
- Tri des colonnes :
  - ✅ **À conserver** : identifiant pseudonymisé, tranche d'âge, sexe, dates d'entrée/sortie, service, codes diagnostics (CIM-10), nombre d'admissions récentes.
  - ❌ **À exclure** : nom, date de naissance exacte, numéro national, profession, noms de soignants, comptes-rendus complets, imageries brutes.
- **Règle d'or :** si une colonne n'a pas d'utilité statistique prouvée, elle ne doit pas être extraite.

## 2. La base légale (Art. 6 RGPD)

| Base | Qui peut l'utiliser |
| :--- | :--- |
| **Mission d'intérêt public** (Art. 6.1.e) | Hôpitaux publics ou privés de service public |
| **Intérêt légitime** (Art. 6.1.f) | Établissements privés, avec une mise en balance des intérêts écrite |

> Le **consentement** (Art. 6.1.a) est déconseillé en milieu hospitalier : patient vulnérable → consentement peu libre, et les refus massifs biaiseraient l'étude.

## 3. La condition supplémentaire pour la santé (Art. 9 RGPD)

Les données de santé sont **« sensibles » : traitement interdit par défaut** (Art. 9.1). Une base légale (Art. 6) ne suffit pas, il faut une **exception** (Art. 9.2) :

- **Art. 9.2.h** — médecine préventive, diagnostic, gestion des services de soins → **condition stricte (Art. 9.3)** : traitement par un professionnel soumis au **secret professionnel**.
- **Art. 9.2.j** — recherche scientifique/statistique (ex : étude rétrospective sur archives) → **condition stricte (Art. 89.1)** : **pseudonymisation** dès le départ, anonymisation dès que possible.

## 4. En résumé : la double serrure juridique

Il faut ouvrir **deux serrures simultanément** :
1. **Art. 6** : une base légale valide (intérêt public ou légitime).
2. **Art. 9** : une exception santé (9.2.h ou 9.2.j) **+** respect de la minimisation (Art. 5).

## 5. Cas pratiques

- ✅ **Scénario conforme** : extraction de 6 colonnes seulement, sous supervision du médecin-chef (secret médical), données pseudonymisées (table de correspondance verrouillée), registre sous intérêt public + 9.2.h → **AUTORISÉ**.
- ❌ **Scénario illégal** : dump brut de 85 colonnes (avec noms, observations libres…) confié à un prestataire externe sans supervision médicale, avec « consentement » pré-coché dans un formulaire → **triple infraction** (Art. 5 + 6 + 9).

# Chapitre 4 : Définir un usage autorisé et proportionné

Pour construire un modèle prédictif efficace sans enfreindre la loi, l'équipe projet doit résister à la tentation du *« prenons toutes les données au cas où »*. En matière de protection des données, la qualité et la pertinence priment toujours sur la quantité.

---

## 1. Le principe de minimisation : Quelles colonnes sont réellement nécessaires ? (Article 5)

L'**article 5.1.c du RGPD** impose le **principe de minimisation des données** : celles-ci doivent être *« adéquates, pertinentes et limitées à ce qui est nécessaire au regard des finalités pour lesquelles elles sont traitées »*.

Demander l'intégralité du dossier patient (comptes-rendus complets, notes psychologiques, antécédents familiaux non liés, adresse postale complète, numéro de registre national/sécurité sociale) est **illégal et disproportionné**.

### Tri des colonnes pour l'étude des réhospitalisations :

| Catégorie | ✅ Colonnes nécessaires (À conserver) | ❌ Colonnes inutiles / risquées (À exclure) |
| :--- | :--- | :--- |
| **Identité** | **Identifiant pseudonymisé** (ex: `PAT_8942`) pour relier les séjours d'un même patient sans connaître son nom. | Nom, prénom, date de naissance exacte, numéro national, adresse, téléphone (inutile pour la phase d'étude). |
| **Profil** | **Tranche d'âge** (ou âge en années), sexe biologique. | Profession, situation familiale détaillée, religion. |
| **Séjour** | **Dates d'entrée et de sortie** (pour calculer la durée du séjour et le délai avant réadmission), service médical concerné, mode de sortie (domicile, maison de repos). | Nom des médecins ou infirmiers traitants. |
| **Médical** | **Codes diagnostics standardisés** (ex: codes CIM-10), comorbidités principales, résultats biologiques clés liés à la pathologie, nombre d'admissions sur les 12 derniers mois. | Notes libres des soignants, comptes-rendus opératoires complets, imageries brutes, dossiers psychiatriques. |

> **Règle d'or :** Si une colonne n'a pas d'utilité statistique prouvée pour prédire la réhospitalisation, elle ne doit pas être extraite de la base de l'hôpital.

---

## 2. Sur quelle base légale mener l'étude ? (Article 6)

Selon l'**article 6 du RGPD**, tout traitement de données personnelles doit reposer sur l'une des **6 bases légales** prévues par le texte. Dans le cadre d'un hôpital souhaitant améliorer la prise en charge et prévenir les réhospitalisations, deux bases principales peuvent être envisagées selon le statut de l'établissement :

1. **La mission d'intérêt public (Article 6.1.e) :** C'est la base privilégiée pour les **hôpitaux publics** (ou privés assurant une mission de service public). La prévention des réhospitalisations participe directement à l'amélioration de la santé publique et de la qualité des soins.
2. **L'intérêt légitime (Article 6.1.f) :** Peut être invoqué par un établissement **privé**, à condition de démontrer par écrit (via une mise en balance des intérêts) que l'amélioration du suivi médical ne porte pas atteinte aux droits et libertés fondamentaux des patients.

*(Note : Le **consentement** [Art. 6.1.a] est généralement déconseillé en milieu hospitalier pour ce type d'étude statistique globale, car le patient, en situation de vulnérabilité face au corps médical, ne dispose pas toujours d'un choix véritablement libre, et des refus massifs fausseraient l'étude).*

---

## 3. La condition supplémentaire pour les données de santé (Article 9)

Les données médicales sont des **données dites « sensibles »** (catégories particulières de données). L'**article 9.1 du RGPD** pose un principe clair : **leur traitement est par défaut strictement interdit.**

Pour lever cette interdiction, il ne suffit pas d'avoir une base légale (Article 6) : il faut **impérativement** remplir l'une des exceptions prévues à l'**article 9.2 du RGPD**.

Pour notre projet, deux exceptions cumulables ou alternatives s'appliquent selon l'étape du projet :

* **Pour l'amélioration de la prise en charge et le suivi médical → Article 9.2.h :**
  Le traitement est autorisé s'il est nécessaire aux fins de la **médecine préventive**, de diagnostics médicaux, ou de la **gestion des systèmes et des services de soins de santé**. 
  * **Condition stricte (Article 9.3) :** Les données doivent être traitées par un professionnel de santé soumis au **secret professionnel** (ou sous sa responsabilité).
* **Pour la phase purement statistique / recherche scientifique → Article 9.2.j :**
  Si l'équipe réalise d'abord une étude rétrospective sur des archives pour tester la faisabilité mathématique du modèle, elle peut s'appuyer sur l'exception liée à la **recherche scientifique ou statistique**.
  * **Condition stricte (Article 89.1) :** L'équipe doit mettre en place des garanties techniques immédiates, au premier rang desquelles figure la **pseudonymisation** (voire l'anonymisation dès que possible).

---

## En résumé : La double serrure juridique

Pour avoir le feu vert légal, l'équipe projet doit déverrouiller **deux serrures simultanément** :

1. **Serrure 1 (Art. 6) :** Avoir une base légale valide (*Mission d'intérêt public* ou *Intérêt légitime*).
2. **Serrure 2 (Art. 9) :** Remplir une exception santé (*Gestion des soins de santé sous secret médical* [9.2.h] ou *Recherche statistique avec pseudonymisation* [9.2.j]) **ET** limiter l'extraction aux seules colonnes indispensables (Art. 5).


---

## 4. Cas pratiques : Minimisation et accès aux données de santé

Pour bien appliquer les articles 5, 6 et 9, voici la différence entre une extraction conforme et une extraction illégale :

### ✅ Exemple 1 : Quand est-ce AUTORISÉ ? (Le scénario conforme)

L'équipe data de l'hôpital travaille sous la supervision du **Médecin-Chef** (soumis au secret professionnel, respectant ainsi l'**Article 9.3**).

1. **Application de la minimisation (Article 5) :** Le service informatique extrait uniquement une vue filtrée contenant **6 colonnes** : un identifiant aléatoire (`ID_9921`), la tranche d'âge (`60-65 ans`), le sexe, le code diagnostic principal (`CIM-10`), la durée du séjour (`4 jours`) et un indicateur binaire de réadmission à 30 jours (`Oui/Non`).
2. **Base légale et exception santé (Articles 6 et 9) :** Le traitement est inscrit au registre de l'hôpital sous la base de la **mission d'intérêt public** (Art. 6.1.e) et de la **gestion des services de soins de santé** (Art. 9.2.h).
3. **Sécurité :** La table de correspondance permettant de retrouver le nom réel du patient derrière `ID_9921` reste verrouillée chez le médecin coordinateur ; les data scientists n'y ont jamais accès.

> **Verdict : AUTORISÉ.** La double serrure juridique (Art. 6 + Art. 9) est ouverte, les données sont pseudonymisées et strictement limitées au besoin de l'étude.

---

### ❌ Exemple 2 : Quand est-ce INTERDIT ? (Le scénario illégal)

Pour gagner du temps, le responsable informatique fait un « dump » (une copie brute) de la base de données des urgences et la remet directement à un prestataire externe chargé de créer l'IA.

1. **Violation de la minimisation (Article 5) :** Le fichier transmis contient **85 colonnes**, incluant les noms, prénoms, numéros de téléphone, adresses, numéros de registre national, ainsi que l'intégralité des **observations libres des médecins** (mentionnant parfois des addictions, des violences conjugales ou des antécédents psychiatriques sans lien avec l'étude).
2. **Violation de l'Article 9 (Données de santé) :** Aucun professionnel soumis au secret médical ne supervise l'accès du prestataire externe, et aucun contrat de sous-traitance encadrant le secret médical n'a été signé.
3. **Détournement de base légale (Article 6) :** L'hôpital prétend s'appuyer sur le consentement des patients, mais la case à cocher était pré-cochée dans le formulaire d'admission aux urgences.

> **Verdict : STRICTEMENT INTERDIT (Triple infraction).**
> * **Violation de l'Art. 5 :** Collecte massive et disproportionnée de données inutiles.
> * **Violation de l'Art. 6 :** Le consentement pré-coché est nul et non avenu.
> * **Violation de l'Art. 9 :** Rupture du secret médical et exposition illégale de données de santé ultra-sensibles.
