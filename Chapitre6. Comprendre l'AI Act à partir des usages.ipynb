# Chapitre 6 : Comprendre l'AI Act à partir des usages

Contrairement à une idée reçue, le règlement européen sur l'intelligence artificielle (**AI Act**) n'interdit pas l'innovation en santé et ne traite pas toutes les technologies de la même manière. Sa philosophie repose sur **une approche fondée sur les risques** : ce n'est pas la technologie en soi qui est réglementée, mais **l'usage prévu** et son impact potentiel sur la santé, la sécurité et les droits fondamentaux des personnes.

---

## 1. La pyramide des risques de l'AI Act

La Commission européenne classe les systèmes d'IA selon **4 niveaux de risque**, du plus faible au plus strict :

1. **Risque inacceptable (Interdit) :** Systèmes menaçant les droits fondamentaux (ex. : notation sociale, manipulation comportementale subliminale).
2. **Haut risque (Strictement encadré) :** Systèmes pouvant affecter gravement la santé, la sécurité ou l'accès aux services essentiels (dont les soins de santé et le triage médical).
3. **Risque spécifique de transparence (Obligation d'information) :** Systèmes interagissant directement avec des humains (ex. : chatbots, génération de contenu) où l'utilisateur doit savoir qu'il parle à une machine.
4. **Risque minimal ou nul (Libre) :** Applications courantes sans impact sur les droits (ex. : filtres anti-spam, optimisation logistique interne).

---

## 2. Comparaison de deux IA dans le même hôpital

Pour comprendre comment l'usage prévu fait basculer un projet d'une catégorie à l'autre, comparons deux outils déployés au sein de notre établissement :

| Critères de comparaison | IA n°1 : Chatbot d'accueil (Horaires de visite) | IA n°2 : Triage aux urgences (Ordre de passage) |
| :--- | :--- | :--- |
| **Usage prévu** | Répondre aux questions pratiques des visiteurs (horaires, plan d'accès, cafétéria). | Évaluer la gravité des symptômes et recommander l'ordre de prise en charge des patients. |
| **Classification AI Act** | **Risque spécifique de transparence** (Risque limité). | **Système d'IA à HAUT RISQUE** (Annexe III – accès aux services de soins essentiels / triage d'urgence, et potentiel dispositif médical). |
| **Impact en cas d'erreur** | **Faible :** Un visiteur arrive 30 minutes trop tôt ou se trompe de bâtiment. Aucun danger vital. | **Critique :** Un retard de prise en charge peut entraîner une aggravation grave, des séquelles ou le décès d'un patient. |
| **Obligations principales** | **Transparence simple (Art. 50) :** Informer clairement l'utilisateur dès le début qu'il interagit avec une IA et non un humain. | **Conformité lourde (Art. 8 à 15) :** Gestion des risques, données d'entraînement sans biais, documentation technique, traçabilité, cybersécurité et **contrôle humain obligatoire**. |

### Ce qu'il faut retenir pour notre projet de réhospitalisation :
Notre futur outil de prédiction des réhospitalisations se rapproche beaucoup plus de **l'IA n°2 (Haut risque)** que du simple chatbot. Parce qu'il influence **qui sera rappelé en priorité par l'équipe médicale**, une erreur de l'algorithme (un faux négatif qui oublie un patient fragile, ou un biais contre une catégorie de patients) a un impact direct sur la santé et l'équité d'accès aux soins.

---

## 3. Quelles obligations concrètes examiner pour une IA à haut risque ?

Dès lors que l'usage prévu classe l'outil dans la catégorie **« Haut risque »**, l'équipe projet (en lien avec le fournisseur de l'outil et le DPO) doit vérifier **5 exigences clés** avant toute mise en service :

1. **Qualité et gouvernance des données (Art. 10) :** Les données utilisées pour entraîner l'IA doivent être représentatives, exactes et exemptes de biais discriminatoires (âge, sexe, origine, lieu de résidence).
2. **Transparence et explicabilité (Art. 13) :** L'outil ne doit pas être une « boîte noire ». Le médecin ou l'infirmière doit pouvoir comprendre *pourquoi* l'IA recommande de rappeler tel patient (ex. : affichage des facteurs de risque principaux).
3. **Contrôle humain effectif / *Human oversight* (Art. 14) :** L'IA est un outil d'**aide à la décision**, jamais un décideur autonome. Un professionnel de santé formé doit pouvoir ignorer, corriger ou annuler la recommandation de l'IA à tout moment.
4. **Robustesse, exactitude et cybersécurité (Art. 15) :** L'IA doit être testée pour résister aux erreurs, aux données aberrantes et aux cyberattaques.
5. **Traçabilité et journalisation (Art. 12) :** Le système doit enregistrer automatiquement son fonctionnement (logs) pour permettre de vérifier *a posteriori* comment une recommandation a été produite en cas d'incident médical.

## 4. Cas pratiques : L'impact de l'usage sur la conformité

### ✅ Exemple 1 : Quand est-ce AUTORISÉ ? (Le scénario conforme)
* **Pour le chatbot d'accueil :** Dès l'ouverture de la fenêtre de discussion sur le site de l'hôpital, un bandeau affiche : *« Bonjour, je suis un assistant virtuel automatisé. Comment puis-je vous aider ? »*.
* **Pour l'IA de suivi / triage :** L'hôpital déploie un modèle validé cliniquement qui affiche un score de priorité accompagné des variables explicatives. Une infirmière de triage (formée aux limites de l'outil) consulte la liste, vérifie le dossier clinique et valide manuellement l'ordre des appels ou des passages. Chaque décision et modification humaine est journalisée.

> **Verdict : AUTORISÉ.** L'obligation de transparence est remplie pour le chatbot, et les garde-fous du haut risque (explicabilité, absence de biais, contrôle humain) sont respectés pour l'outil médical.

---

### ❌ Exemple 2 : Quand est-ce INTERDIT ? (Le scénario illégal)
* **Pour le chatbot d'accueil :** L'hôpital utilise une voix de synthèse ultra-réaliste au téléphone qui se fait passer pour une secrétaire humaine (« Bonjour, ici Claire de l'accueil ») sans jamais mentionner qu'il s'agit d'une IA.
* **Pour l'IA de suivi / triage :** L'hôpital achète un algorithme « boîte noire » entraîné uniquement sur une population jeune et urbaine, et le configure pour **classer automatiquement les patients sans possibilité pour l'infirmière de modifier l'ordre** (ou en pénalisant les patients âgés pour « optimiser la rotation des lits »).

> **Verdict : STRICTEMENT INTERDIT.**
> * **Violation de l'obligation de transparence (Art. 50) :** Tromperie sur la nature artificielle de l'interlocuteur.
> * **Violation des règles sur le haut risque (Art. 10, 13 et 14) :** Biais discriminatoire grave, opacité totale du modèle et suppression du contrôle humain sur une décision vitale.
