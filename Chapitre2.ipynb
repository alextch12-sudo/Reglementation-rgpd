# Chapitre 2 — Reconnaître une donnée personnelle (et une donnée de santé)

## 🎯 Objectifs pédagogiques

À l'issue de ce chapitre, l'équipe projet saura :

1. **Distinguer** une donnée directement identifiante d'une donnée indirectement identifiante.
2. **Comprendre** le risque de réidentification par recoupement d'informations.
3. **Expliquer** pourquoi un diagnostic bénéficie d'un statut juridique renforcé en tant que donnée de santé.
4. **Appliquer** le réflexe de minimisation avant même d'explorer les dossiers patients.

---

## 1. Étude de cas : analyse d'un dossier patient fictif

Imaginons que l'équipe récupère un extrait de dossier patient contenant uniquement les quatre informations suivantes :

| Champ | Valeur fictive |
|---|---|
| **Nom** | Mme Claire Dubois |
| **Âge** | 84 ans |
| **Commune** | Châtelet |
| **Diagnostic** | Insuffisance cardiaque décompensée (réhospitalisation à 15 jours) |

> Avant d'utiliser ces données pour entraîner ou tester un outil d'IA, posons-nous **deux questions essentielles** au sens du Règlement Général sur la Protection des Données (RGPD).

---

## 2. Quelles informations permettent de reconnaître une personne, seules ou par recoupement ?

Selon l'**article 4 du RGPD**, une donnée à caractère personnel est *toute information se rapportant à une personne physique identifiée ou identifiable*. Une personne peut être reconnue de deux manières :

### A. L'identification directe (seule)

- **Le nom** (*Mme Claire Dubois*) : c'est un **identifiant direct**. À lui seul, il permet de désigner nommément une personne physique.

### B. L'identification indirecte (par recoupement)

Supposons que l'équipe décide de **supprimer la colonne `Nom`** en pensant que le dossier devient « anonyme ». **C'est une erreur fréquente.**

Les trois informations restantes, combinées entre elles, permettent de retrouver la personne par recoupement (ou *inférence*) :

- **L'âge** (84 ans) **+ la commune** (Châtelet) : réduisent déjà considérablement le nombre d'habitants correspondants dans une commune de taille moyenne.
- **L'ajout du diagnostic** (insuffisance cardiaque décompensée) — *et a fortiori* la date d'admission ou de sortie associée au dossier hospitalier : permettent de cibler **quasi certainement une personne unique** au sein de l'hôpital ou de la commune, en particulier pour une pathologie précise ou dans une petite localité.

> ### 💡 À retenir pour l'équipe
> Supprimer le nom et le prénom **ne rend pas un dossier anonyme** : il s'agit seulement de **pseudonymisation** (ou de suppression d'identifiants directs).
> Tant qu'il est raisonnablement possible de réidentifier un patient en croisant l'âge, le lieu de résidence et le contexte médical, le dossier reste soumis à **l'intégralité des règles du RGPD**.

---

## 3. Qu'est-ce qui fait du diagnostic une donnée de santé ?

Au sein de ce dossier, le diagnostic a un statut particulier : c'est une **donnée concernant la santé** (article 4.15 du RGPD), classée parmi les **catégories particulières de données** — dites **données sensibles** (article 9 du RGPD).

### Trois critères en font une donnée de santé

| # | Critère | Explication |
|---|---|---|
| 1 | **Nature médicale directe** | Le diagnostic révèle explicitement l'état de santé physique ou mentale passé, présent ou futur d'une personne (pathologie, fragilité clinique, risque médical). |
| 2 | **Lien avec la prise en charge de soins** | Il est produit dans un cadre médical et renseigne sur le parcours de soins du patient (hospitalisation, suivi, traitement). |
| 3 | **Risque élevé pour les droits et libertés** | La divulgation ou le mauvais usage d'un diagnostic peut entraîner des conséquences graves (atteinte à l'intimité de la vie privée, stigmatisation, discrimination par des assureurs ou employeurs, anxiété). |

### ⚠️ Conséquence pratique pour notre projet

Parce que le diagnostic est une **donnée sensible** :

- Son traitement est **interdit par principe** par le RGPD, sauf à entrer dans une **exception légale stricte** (par exemple la prise en charge sanitaire, la gestion des services de soins ou l'intérêt public dans le domaine de la santé, sous le contrôle de professionnels soumis au secret médical).
- L'équipe technique qui conçoit l'IA **ne doit pas accéder librement aux diagnostics nominatifs** :
  - **Cloisonner** les accès ;
  - Travailler sur des **variables strictement nécessaires** (par exemple des tranches d'âge plutôt que l'âge exact, un code postal élargi ou aucun lieu précis si inutile) ;
  - **Associer le DPO** (Délégué à la Protection des Données) **dès la conception**.
