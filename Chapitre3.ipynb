# 🏥 Chapitre 3 — Pseudonymiser n'est pas anonymiser

---

## 🎯 Situation

L'hôpital remplace les noms des patients par des codes :

| Avant         | Après    |
| ------------- | -------- |
| Marie Dupont  | PAT-001  |
| Jean Martin   | PAT-002  |
| Sophie Leroy  | PAT-003  |

L'hôpital conserve de son côté une **table de correspondance** qui relie chaque code à l'identité réelle. Puis il transmet le fichier à une équipe d'analyse.

> ### 👉 Le fichier est-il devenu anonyme ?
>
> ## ❌ NON
>
> Selon la **CNIL**, remplacer les noms par des codes est une **pseudonymisation**, pas une anonymisation. Les données restent donc des **données personnelles** : il existe toujours un moyen de retrouver l'identité des personnes.
>
> 🔑 *Une donnée est anonymisée uniquement lorsqu'il est devenu impossible, en pratique et de manière irréversible, d'identifier une personne.*

---

## 🕵️ Le piège : reconnaître quelqu'un sans son nom

Même sans la table de correspondance, certaines informations peuvent permettre de retrouver une personne. La CNIL rappelle qu'une personne peut être identifiée **non seulement par un nom, mais aussi par recoupement** de plusieurs informations.

### 🔎 Exemple 1 — une personne très âgée

| Code patient | Âge     | Commune   |
| ------------ | ------- | --------- |
| PAT-267      | 103 ans | Charleroi |

Si une seule personne de 103 ans vit dans cette commune, son identité devient facile à deviner.

> ➡️ **Risque de ré-identification** — la CNIL cite les *valeurs rares* (par exemple les personnes centenaires) comme un facteur important de ré-identification.

### 🔎 Exemple 2 — maladie rare + lieu

| Code patient | Maladie                  | Commune        |
| ------------ | ------------------------ | -------------- |
| PAT-104      | Maladie génétique rare   | Petite commune |

Même sans connaître le nom, beaucoup de personnes sauront de qui il s'agit.

> ➡️ **Risque de ré-identification** par combinaison d'informations atypiques.

### 🔎 Exemple 3 — accumulation de détails

| Code patient | Date de naissance | Sexe  | Commune |
| ------------ | ----------------- | ----- | ------- |
| PAT-521      | 14/01/1972        | Femme | Fleurus |

Une personne disposant d'autres sources d'information (registres, réseaux, connaissances locales) pourrait retrouver le patient.

> ➡️ **Croisement possible** avec des données extérieures.

---

## 🧠 La bonne question à se poser

| ❌ Mauvais réflexe                          | ✅ Réflexe CNIL                                                                                          |
| ------------------------------------------- | -------------------------------------------------------------------------------------------------------- |
| « Avons-nous supprimé le nom ? »            | « Peut-on encore reconnaître ou retrouver un patient grâce aux informations restantes, seules ou croisées ? » |

---

## ✅ vs ❌ : comparer deux jeux de données

### ✅ Données réellement anonymisées

| Tranche d'âge | Région  | Nombre de patients |
| ------------- | ------- | ------------------ |
| 60-70 ans     | Hainaut | 1 250              |
| 70-80 ans     | Hainaut | 980                |

✔️ Pas de nom · ✔️ Pas de date de naissance · ✔️ Pas d'identifiant individuel · ✔️ Données agrégées sur des groupes suffisamment larges

➡️ Il devient extrêmement difficile de rattacher l'information à une personne précise.

### ❌ Données seulement pseudonymisées

| Code    | Âge     | Commune  | Pathologie     |
| ------- | ------- | -------- | -------------- |
| PAT-001 | 101 ans | Châtelet | Pathologie rare |

Le nom a disparu, **mais** :

| Élément conservé                         | Conséquence                          |
| ---------------------------------------- | ------------------------------------ |
| ✅ Âge exact conservé                     | Combiné à la commune → cible unique  |
| ✅ Commune conservée                      | Réduit fortement le nombre depersonnes |
| ✅ Pathologie conservée                   | information de santé sensible        |
| ✅ Table de correspondance existante      | L'identifiant reste réversible       |

➡️ On n'a **pas anonymisé** : on a seulement **pseudonymisé**. Le RGPD continue donc de s'appliquer.

---

## 🎓 À retenir

| | Pseudonymiser 🔒 | Anonymiser 🗑️ |
| --- | --- | --- |
| Ce que c'est | **Cacher** le nom (ou le remplacer par un code) | Rendre l'identification **impossible**, même en croisant les données |
| Table de correspondance | Existe (détention possible) | Aucun lien ne subsiste |
| Statut des données | Toujours des **données personnelles** → RGPD applicable | Plus des données personnelles → RGPD hors champ |

> 💡 **Test simple** : si quelqu'un peut encore dire *« je sais de quel patient il s'agit »* grâce aux données restantes, les données ne sont probablement pas anonymes.
