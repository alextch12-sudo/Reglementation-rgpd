

# Chapitre 8 — Appliquer les notions à un projet (Cas pratique)

## 🎯 Objectifs pédagogiques

À l'issue de ce chapitre de synthèse, l'équipe projet saura :

1. **Formuler** une recommandation argumentée avant le lancement d'un test d'IA en milieu hospitalier.
2. **Définir** le périmètre exact des données utilisables (minimisation et pseudonymisation).
3. **Identifier** les vérifications critiques à mener avant de faire confiance aux prédictions du modèle.
4. **Lister** les informations manquantes nécessaires pour sécuriser juridiquement et cliniquement la décision.

---

## 1. Mise en situation : le projet pilote de l'hôpital

La direction de l'hôpital souhaite lancer une phase de test (*« pilote »*) : utiliser un outil d'intelligence artificielle pour **analyser les dossiers de sortie** et **choisir quels patients rappeler** afin de prévenir les réhospitalisations.

> On demande à notre équipe projet un avis clair : *« Que nous conseillez-vous avant de démarrer ce test ? »*

---

## 2. Notre recommandation : un « Oui, sous conditions strictes » (phase pilote encadrée)

Nous conseillons à l'hôpital de **ne pas déployer l'IA directement en décision automatique**, mais d'autoriser uniquement un **test en conditions contrôlées** (*« mode observation »* ou *aide à la décision supervisée*), après validation du cadre juridique et technique.

> ### 💡 À retenir
> Concrètement, l'IA ne doit pas « choisir » seule les patients : elle doit **proposer un score ou une liste de priorisation** que l'équipe soignante **examine et valide** avant tout appel.

---

## 3. Sur quelles données l'équipe peut-elle s'appuyer ? (Principe de minimisation)

Pour respecter le RGPD (notamment le **statut sensible des données de santé** à l'article 9 et le **principe de minimisation**), l'équipe technique ne doit pas entraîner ni tester le modèle sur l'intégralité brute des dossiers médicaux.

### ✅ Données autorisées pour le test (strictement nécessaires et pseudonymisées)

- **Un identifiant technique pseudonymisé** (ex. `PAT_8942`) : la table de correspondance avec le nom et le numéro de téléphone reste **cloisonnée** et accessible uniquement aux soignants chargés des rappels.
- **Des données cliniques ciblées et pertinentes** : motif principal d'hospitalisation (catégorie diagnostique), durée de séjour, antécédents d'admissions récentes, présence de comorbidités majeures ou de polymédication.
- **Des données contextuelles généralisées** : tranche d'âge (ex. *80-89 ans* plutôt que la date de naissance exacte), secteur géographique élargi (pour limiter le risque de réidentification par recoupement).

### ❌ Données à exclure du jeu de test

- **Les identifiants directs** : nom, prénom, adresse postale précise, numéro de téléphone, numéro de registre national / sécurité sociale.
- **Les notes cliniques libres non nettoyées** : comptes-rendus complets contenant des détails personnels ou familiaux sans lien direct avec le risque de réhospitalisation.

---

## 4. Quelle vérification indispensable faire avant de se fier aux prédictions ?

Avant d'utiliser les suggestions de l'IA pour organiser les appels des soignants, l'équipe doit impérativement mener une **validation clinique et statistique**, accompagnée d'un **contrôle des biais** et d'une **supervision humaine** (exigences clés de l'*AI Act*) :

### A. Vérifier la fiabilité et l'équilibre des erreurs (faux négatifs vs faux positifs)

- **Tester le modèle** sur des données historiques déjà connues (*évaluation rétrospective*) ou en **mode silencieux** (l'IA calcule un risque sans influencer les appels, puis on compare ses prédictions à la réalité clinique).
- En santé, un **faux négatif** (un patient à haut risque que l'IA ne signale pas) est **critique** car il prive un patient fragile d'un appel préventif.

### B. Audit des biais (équité d'accès aux soins)

- Vérifier que l'IA **ne sous-estime pas le risque** pour certains groupes de patients (par exemple selon l'âge, le sexe, le lieu de résidence ou une moindre fréquentation passée des soins).

### C. Tester le protocole de contrôle humain

- S'assurer que les infirmiers ou médecins disposent d'**indicateurs compréhensibles** (*explicabilité* : *pourquoi ce patient est-il signalé ?*) et gardent la **liberté d'ajouter ou de retirer** des patients selon leur jugement clinique.

---

## 5. Ce qu'il nous manque pour trancher définitivement

Pour passer d'un avis de principe à un **feu vert opérationnel**, l'équipe a besoin de lever **quatre zones d'ombre** avec la direction, les soignants et le DPO :

| Dimension | Question ouverte / Élément manquant | Interlocuteur clé |
|---|---|---|
| **Cadre RGPD** | Quelle est la base légale retenue, comment les patients sont-ils informés et l'AIPD (Analyse d'Impact relative à la Protection des Données) a-t-elle été validée ? | DPO (Délégué à la Protection des Données) |
| **Qualification AI Act** | S'agit-il d'un outil interne développé par l'hôpital (*fournisseur*) ou d'un logiciel acheté (*déployeur*) ? L'outil relève-t-il d'un dispositif médical ou d'un système à haut risque ? | Direction juridique / Qualité & Risques |
| **Capacité opérationnelle** | Combien d'appels l'équipe soignante peut-elle réellement passer par jour, et quel seuil d'alerte évite la surcharge tout en captant les cas graves ? | Cadres de santé / Équipe médicale |
| **Qualité des données sources** | Les dossiers historiques de l'hôpital sont-ils suffisamment complets, standardisés et exempts d'erreurs de saisie pour entraîner ou évaluer le modèle ? | DIM / DSI |

---

## 6. 📋 Synthèse de décision (Fiche réflexe équipe)

> **Décision proposée** : Lancer un **pilote en mode supervisé** sur des données **pseudonymisées et minimisées**, précédé d'une **évaluation sur historique** (et **audit des biais**), sous réserve de :
>
> - la **validation de l'AIPD** avec le DPO ;
> - la **clarification du statut réglementaire** (AI Act / dispositif médical) de l'outil.
````

**Choix de mise en page, cohérents avec le Chapitre 2 :**

- Même hiérarchie de titres et séparateurs `---` entre les grandes sections.
- Section 3 scindée en **✅ autorisées / ❌ à exclure** — lecture immédiate de l'opposition.
- Section 4 structurée en trois sous-parties **A / B / C**, comme l'opposition direct/indirect du chapitre 2.
- Le tableau des « zones d'ombre » conservé (il était déjà tabulaire), avec les interlocuteurs abrégés (*DIM / DSI*) pour alléger.
- La synthèse finale-presentée en **bloc de citation** avec les deux conditions en liste, pour servir de « fiche réflexe » visuellement détachable.


# Chapitre 8 — Appliquer les notions à un projet (Cas pratique)

## 🎯 Objectifs pédagogiques

À l'issue de ce chapitre de synthèse, l'équipe projet saura :

1. **Formuler** une recommandation argumentée avant le lancement d'un test d'IA en milieu hospitalier.
2. **Définir** le périmètre exact des données utilisables (minimisation et pseudonymisation).
3. **Identifier** les vérifications critiques à mener avant de faire confiance aux prédictions du modèle.
4. **Lister** les informations manquantes nécessaires pour sécuriser juridiquement et cliniquement la décision.

---

## 1. Mise en situation : le projet pilote de l'hôpital

La direction de l'hôpital souhaite lancer une phase de test (*« pilote »*) : utiliser un outil d'intelligence artificielle pour **analyser les dossiers de sortie** et **choisir quels patients rappeler** afin de prévenir les réhospitalisations.

> On demande à notre équipe projet un avis clair : *« Que nous conseillez-vous avant de démarrer ce test ? »*

---

## 2. Notre recommandation : un « Oui, sous conditions strictes » (phase pilote encadrée)

Nous conseillons à l'hôpital de **ne pas déployer l'IA directement en décision automatique**, mais d'autoriser uniquement un **test en conditions contrôlées** (*« mode observation »* ou *aide à la décision supervisée*), après validation du cadre juridique et technique.

> ### 💡 À retenir
> Concrètement, l'IA ne doit pas « choisir » seule les patients : elle doit **proposer un score ou une liste de priorisation** que l'équipe soignante **examine et valide** avant tout appel.

---

## 3. Sur quelles données l'équipe peut-elle s'appuyer ? (Principe de minimisation)

Pour respecter le RGPD (notamment le **statut sensible des données de santé** à l'article 9 et le **principe de minimisation**), l'équipe technique ne doit pas entraîner ni tester le modèle sur l'intégralité brute des dossiers médicaux.

### ✅ Données autorisées pour le test (strictement nécessaires et pseudonymisées)

- **Un identifiant technique pseudonymisé** (ex. `PAT_8942`) : la table de correspondance avec le nom et le numéro de téléphone reste **cloisonnée** et accessible uniquement aux soignants chargés des rappels.
- **Des données cliniques ciblées et pertinentes** : motif principal d'hospitalisation (catégorie diagnostique), durée de séjour, antécédents d'admissions récentes, présence de comorbidités majeures ou de polymédication.
- **Des données contextuelles généralisées** : tranche d'âge (ex. *80-89 ans* plutôt que la date de naissance exacte), secteur géographique élargi (pour limiter le risque de réidentification par recoupement).

### ❌ Données à exclure du jeu de test

- **Les identifiants directs** : nom, prénom, adresse postale précise, numéro de téléphone, numéro de registre national / sécurité sociale.
- **Les notes cliniques libres non nettoyées** : comptes-rendus complets contenant des détails personnels ou familiaux sans lien direct avec le risque de réhospitalisation.

---

## 4. Quelle vérification indispensable faire avant de se fier aux prédictions ?

Avant d'utiliser les suggestions de l'IA pour organiser les appels des soignants, l'équipe doit impérativement mener une **validation clinique et statistique**, accompagnée d'un **contrôle des biais** et d'une **supervision humaine** (exigences clés de l'*AI Act*) :

### A. Vérifier la fiabilité et l'équilibre des erreurs (faux négatifs vs faux positifs)

- **Tester le modèle** sur des données historiques déjà connues (*évaluation rétrospective*) ou en **mode silencieux** (l'IA calcule un risque sans influencer les appels, puis on compare ses prédictions à la réalité clinique).
- En santé, un **faux négatif** (un patient à haut risque que l'IA ne signale pas) est **critique** car il prive un patient fragile d'un appel préventif.

### B. Audit des biais (équité d'accès aux soins)

- Vérifier que l'IA **ne sous-estime pas le risque** pour certains groupes de patients (par exemple selon l'âge, le sexe, le lieu de résidence ou une moindre fréquentation passée des soins).

### C. Tester le protocole de contrôle humain

- S'assurer que les infirmiers ou médecins disposent d'**indicateurs compréhensibles** (*explicabilité* : *pourquoi ce patient est-il signalé ?*) et gardent la **liberté d'ajouter ou de retirer** des patients selon leur jugement clinique.

---

## 5. Ce qu'il nous manque pour trancher définitivement

Pour passer d'un avis de principe à un **feu vert opérationnel**, l'équipe a besoin de lever **quatre zones d'ombre** avec la direction, les soignants et le DPO :

| Dimension | Question ouverte / Élément manquant | Interlocuteur clé |
|---|---|---|
| **Cadre RGPD** | Quelle est la base légale retenue, comment les patients sont-ils informés et l'AIPD (Analyse d'Impact relative à la Protection des Données) a-t-elle été validée ? | DPO (Délégué à la Protection des Données) |
| **Qualification AI Act** | S'agit-il d'un outil interne développé par l'hôpital (*fournisseur*) ou d'un logiciel acheté (*déployeur*) ? L'outil relève-t-il d'un dispositif médical ou d'un système à haut risque ? | Direction juridique / Qualité & Risques |
| **Capacité opérationnelle** | Combien d'appels l'équipe soignante peut-elle réellement passer par jour, et quel seuil d'alerte évite la surcharge tout en captant les cas graves ? | Cadres de santé / Équipe médicale |
| **Qualité des données sources** | Les dossiers historiques de l'hôpital sont-ils suffisamment complets, standardisés et exempts d'erreurs de saisie pour entraîner ou évaluer le modèle ? | DIM / DSI |

---

## 6. 📋 Synthèse de décision (Fiche réflexe équipe)

> **Décision proposée** : Lancer un **pilote en mode supervisé** sur des données **pseudonymisées et minimisées**, précédé d'une **évaluation sur historique** (et **audit des biais**), sous réserve de :
>
> - la **validation de l'AIPD** avec le DPO ;
> - la **clarification du statut réglementaire** (AI Act / dispositif médical) de l'outil.
````
