# Chapitre 1 : Deux règlements pour deux questions

Avant de manipuler la moindre donnée médicale ou de déployer un algorithme, l'équipe projet doit comprendre que son initiative est encadrée par deux textes européens majeurs et complémentaires : le **RGPD** et l'**AI Act**. 

Plutôt que de les voir comme des freins, il faut les utiliser comme deux boussoles qui posent chacune une question fondamentale à deux étapes distinctes du projet.

---

## 1. Étape 1 : L'analyse des dossiers patients → La question du RGPD

Dès que l'équipe envisage d'accéder aux dossiers médicaux pour les explorer, les nettoyer ou entraîner un modèle, elle entre dans le champ d'application du **Règlement Général sur la Protection des Données (RGPD)**. Ici, l'objet protégé est **la donnée personnelle** (et plus particulièrement la donnée de santé, considérée comme sensible par l'article 9).

> **La question centrale du RGPD :** 
> *« Avons-nous le droit d'utiliser ces données personnelles de santé, dans quel but précis, et comment garantissons-nous la sécurité et les droits des patients ? »*

Concrètement, cette question oblige l'hôpital à vérifier plusieurs prérequis avant tout traitement :
* **La licéité et la finalité :** Peut-on réutiliser des données collectées initialement pour le soin (finalité primaire) afin de créer un outil de prédiction de réhospitalisation (finalité secondaire) ? Sur quelle base légale ?
* **La minimisation :** Avons-nous réellement besoin de l'intégralité du dossier patient, ou seulement de quelques variables pertinentes (âge, pathologie, durée du séjour) ?
* **La transparence et les droits :** Les patients sont-ils informés de cette utilisation et peuvent-ils s'y opposer ?

---

## 2. Étape 2 : L'utilisation d'une IA pour suggérer qui rappeler → La question de l'AI Act

Une fois les données préparées, l'équipe souhaite utiliser un système d'intelligence artificielle pour évaluer le risque de réhospitalisation et prioriser les appels de suivi. C'est ici qu'intervient le **Règlement européen sur l'Intelligence Artificielle (AI Act)**. Ici, l'objet encadré n'est plus la donnée brute, mais **le système d'IA lui-même** et son impact sur les personnes.

> **La question centrale de l'AI Act :** 
> *« Quel niveau de risque cet outil d'IA présente-t-il pour la santé, la sécurité et les droits fondamentaux des patients, et quelles garanties techniques et humaines devons-nous mettre en place pour le rendre fiable ? »*

Concrètement, cette question amène l'équipe à évaluer :
* **La classification du risque :** Un outil qui oriente la prise en charge médicale ou le suivi des patients est susceptible d'être classé comme un **système d'IA à haut risque** (notamment s'il est considéré comme un dispositif médical ou s'il influence l'accès aux soins).
* **La qualité et l'absence de biais :** L'IA risque-t-elle de discriminer certains profils de patients (ex. : sous-estimer le risque chez les personnes âgées ou certaines catégories socio-économiques) ?
* **La supervision humaine (*Human-in-the-loop*) :** L'IA ne fait que *suggérer*. Comment garantir que le personnel soignant garde le contrôle final et comprenne pourquoi l'outil recommande de rappeler tel patient plutôt qu'un autre (explicabilité) ?

---

## En résumé : Deux boucliers complémentaires

| Dimension | Étape 1 : Les données | Étape 2 : L'algorithme |
| :--- | :--- | :--- |
| **Règlement applicable** | **RGPD** | **AI Act** |
| **Ce qui est protégé** | La vie privée et les données de santé du patient | La sécurité, la santé et les droits fondamentaux face à la machine |
| **Question clé** | *« Est-ce légal et proportionné d'utiliser ces dossiers ? »* | *« Cet outil est-il sûr, transparent, sans biais et sous contrôle humain ? »* |

## 3. Cas pratiques : Ce qui est autorisé vs Ce qui est interdit

Pour bien visualiser la frontière entre un projet conforme et une infraction, voici deux scénarios concrets appliqués à notre hôpital :

### ✅ Exemple 1 : Quand est-ce AUTORISÉ ? (Le scénario conforme)

L'hôpital met en place un protocole strict avant de lancer l'outil :

1. **Côté RGPD (Les données) :** 
   * Le DPO (Délégué à la Protection des Données) valide le traitement sur la base des **soins de santé / mission d'intérêt public** (Art. 6 et Art. 9.2.h du RGPD) et réalise une **Analyse d'Impact (AIPD)**.
   * Seules les variables strictement nécessaires (âge, motif d'admission, résultats biologiques clés) sont extraites et **pseudonymisées** pour entraîner le modèle.
   * Les patients reçoivent une note d'information claire dès leur admission et peuvent s'opposer à ce suivi.
2. **Côté AI Act (L'outil) :** 
   * L'IA est conçue comme un simple **outil d'aide à la décision** (encadré en tant que système à haut risque).
   * Elle affiche un score de risque accompagné de ses facteurs explicatifs (ex. : *"Risque élevé car tension instable à la sortie"*).
   * Une infirmière de coordination examine chaque suggestion et décide seule s'il faut appeler le patient ou non.

> **Verdict : AUTORISÉ.** La donnée est protégée, le patient est informé et le personnel soignant garde le contrôle total sur la décision médicale.

---

### ❌ Exemple 2 : Quand est-ce INTERDIT ? (Le scénario illégal)

Un membre de l'équipe veut aller vite et fait un test « en local » sans validation préalable :

1. **Côté RGPD (Les données) :** 
   * Il exporte l'intégralité des dossiers médicaux (y compris les notes confidentielles et les coordonnées complètes) sur son ordinateur portable personnel non chiffré pour entraîner une IA via un service cloud public non certifié pour la santé.
   * Aucun patient n'a été informé de cette réutilisation de ses données.
2. **Côté AI Act (L'outil) :** 
   * Le système est paramétré pour **envoyer automatiquement des convocations** sans validation humaine.
   * L'algorithme intègre des critères discriminatoires (ex. : baisser le score de priorité des patients vivant dans certains codes postaux jugés « moins enclins à répondre »).

> **Verdict : STRICTEMENT INTERDIT (Double infraction).**
> * **Violation du RGPD :** Détournement de finalité, absence de base légale, défaut de sécurité sur des données sensibles et non-respect du droit à l'information.
> * **Violation de l'AI Act :** Biais discriminatoire (notation géographique/sociale) et prise de décision automatisée affectant la santé sans supervision humaine.
