7. Évaluer une IA au-delà de sa performance globale
Sur 100 patients, 10 sont réhospitalisés. Un modèle qui prédit « aucune réhospitalisation » a donc raison 90 fois. Combien de patients à rappeler a-t-il repérés ? Cherchez un indicateur plus utile pour ce suivi et expliquez comment vérifier si le modèle manque davantage de patients dans une tranche d'âge.


# 🏥 Question 7 : Une IA à 90 % peut-elle être mauvaise ?

## 🎯 Situation

Sur 100 patients :

- 90 ne seront pas réhospitalisés
- 10 seront réhospitalisés

L'IA prédit :

> "Aucun patient ne sera réhospitalisé."

---

## 📊 Résultat

✅ 90 prédictions correctes

❌ 10 prédictions incorrectes

**Précision globale :**

```text
90 / 100 = 90 %
```

Cela semble excellent...

---

## 🚨 Mais combien de patients à risque a-t-elle trouvés ?

L'objectif est d'identifier les patients à rappeler.

Or l'IA a détecté :

```text
0 patient sur 10 à risque
```

➡️ Aucun patient à rappeler n'a été repéré.

---

## 📏 Un indicateur plus utile : le rappel (Recall)

Le rappel répond à la question :

> **"Parmi les patients réellement à risque, combien ont été détectés ?"**

```text
Rappel = Patients à risque détectés
          ÷
          Patients à risque réels
```

Dans notre cas :

```text
0 ÷ 10 = 0 %
```

🚨 Malgré une précision de 90 %, le rappel est de **0 %**.

---

## 👵 L'IA oublie-t-elle davantage certains patients ?

Il faut vérifier les performances par tranche d'âge.

| Tranche d'âge | Patients à risque | Détectés | Rappel |
|---------------|------------------|-----------|---------|
| Moins de 65 ans | 10 | 8 | 80 % |
| Plus de 80 ans | 10 | 3 | 30 % |

👉 L'IA manque davantage de patients chez les plus de 80 ans.

---

## ✅ Exemple CORRECT

L'équipe mesure :

- la précision globale ;
- le rappel des patients à risque ;
- le rappel par tranche d'âge.

✅ Le modèle détecte correctement les patients à risque dans tous les groupes.

---

## ❌ Exemple INCORRECT

L'équipe regarde uniquement :

```text
Précision globale = 90 %
```

et conclut :

> "Le modèle est bon."

❌ Sans vérifier combien de patients à risque ont été oubliés.

---

---

# 🎓 À retenir

> **Une précision globale élevée ne signifie pas qu'une IA est utile.**

Pour un système qui doit identifier des patients à rappeler :

✅ vérifier le **rappel (recall)**

✅ mesurer combien de patients à risque sont manqués

✅ comparer les résultats entre tranches d'âge

✅ rechercher d'éventuelles inégalités de performance

La bonne question n'est donc pas :

> "Combien de fois l'IA a-t-elle raison ?"

mais :

> **"Combien de patients à risque a-t-elle réellement réussi à repérer, pour chaque groupe de patients ?"**
